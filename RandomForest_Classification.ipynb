```python
# Importing requiered libraries 
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import confusion_matrix, classification_report
```

```python
# Loading the dataset from Packt GitHub repository
file_url = (
    'https://raw.githubusercontent.com/'
    'PacktWorkshops/'
    'The-Applied-Artificial-Intelligence-Workshop/'
    'master/Datasets/car.csv'
)

df = pd.read_csv(file_url)

# Encoding all categorical columns
le = LabelEncoder()
for col in df.columns:
    df[col] = le.fit_transform(df[col])

# Separating features and target
label = df.pop("class")
features = df

# Train-test split (same 90/10 split as in activity 4.01)
features_train, features_test, label_train, label_test = train_test_split(
    features, label, test_size=0.10, random_state=42
)

features_train.head()
```

```python
# Creating and Training a RandomForestClassifier
random_forest_classifier = RandomForestClassifier(
    n_estimators=100,
    max_depth=6,
    random_state=168
) 

# Training the model
random_forest_classifier.fit(features_train, label_train) 
```
